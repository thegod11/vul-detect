```python
def sparse_mx_to_torch_sparse_tensor(sparse_mx):
    """Convert a scipy sparse matrix to a torch sparse tensor."""
    sparse_mx = sparse_mx.tocoo().astype(np.float32)
    indices = torch.from_numpy(
        np.vstack((sparse_mx.row, sparse_mx.col)).astype(np.int64))
    values = torch.from_numpy(sparse_mx.data)
    shape = torch.Size(sparse_mx.shape)
    return torch.sparse.FloatTensor(indices, values, shape)
```

```python
def normalize_adj(row, col):
    """Symmetrically normalize adjacency matrix."""
    data = torch.ones(row.size(1))
    adj = sp.coo_matrix((data, (row, col)), shape=(het_graph.num_nodes(), het_graph.num_nodes(),))
    rowsum = np.array(adj.sum(1))

    d_inv_sqrt = np.power(rowsum, -0.5).flatten()
    d_inv_sqrt[np.isinf(d_inv_sqrt)] = 0.
    d_mat_inv_sqrt = sp.diags(d_inv_sqrt)
    return adj.dot(d_mat_inv_sqrt).transpose().dot(d_mat_inv_sqrt).tocoo()
```

```python
def normalize_adj(row, col):
    """Symmetrically normalize adjacency matrix."""
    data = torch.ones(row.size(0))

    adj = sp.coo_matrix((data, (row, col)), shape=(het_graph.num_nodes(), het_graph.num_nodes(),))
    rowsum = np.array(adj.sum(1))

    d_inv_sqrt = np.power(rowsum, -0.5).flatten()
    d_inv_sqrt[np.isinf(d_inv_sqrt)] = 0.
    d_mat_inv_sqrt = sp.diags(d_inv_sqrt)
    return adj.dot(d_mat_inv_sqrt).transpose().dot(d_mat_inv_sqrt).tocoo()
```

```python
def re_featuresv2(adj, features, K, start_hops=0):
    nodes_features = torch.empty(features.shape[0], 1, K + 1 - start_hops, features.shape[1])

    for i in range(features.shape[0]):  # self
        nodes_features[i, 0, 0, :] = features[i]

    x = features + torch.zeros_like(features)

    for i in range(K):  # 0:1-hop，1：2:-hop....k-1:K-hop

        x = torch.matmul(adj, x)
        if i >= start_hops:
            for index in range(features.shape[0]):
                nodes_features[index, 0, i + 1 - start_hops, :] = x[index]

    nodes_features = nodes_features.squeeze()
    return nodes_features  # size= (N, 1, K+1-start_hops, d )
```

```python
import torch
from collections import defaultdict, deque

def compute_deepest_path_pdg(nodes: torch.Tensor, edges: tuple) -> int:
    # 解析边结构
    edge_list = []
    u_tensor, v_tensor = edges  # edges 是 (tensor_u, tensor_v)
    for u, v in zip(u_tensor, v_tensor):
        edge_list.append((u.item(), v.item()))
    
    # 构建邻接表
    adj = defaultdict(list)
    for u, v in edge_list:
        adj[u].append(v)
    
    # Tarjan算法找SCC
    index = 0
    indices = {}
    low_link = {}
    on_stack = set()
    stack = []
    sccs = []
    
    def strongconnect(v):
        # print(v)
        nonlocal index
        indices[v] = index
        low_link[v] = index
        index += 1
        stack.append(v)
        on_stack.add(v)
        
        for w in adj[v]:
            if w not in indices:
                strongconnect(w)
                low_link[v] = min(low_link[v], low_link[w])
            elif w in on_stack:
                low_link[v] = min(low_link[v], indices[w])
        
        if low_link[v] == indices[v]:
            scc = []
            while True:
                w = stack.pop()
                on_stack.remove(w)
                scc.append(w)
                if w == v:
                    break
            sccs.append(scc)
    
    for v in list(adj.keys()):
        if v not in list(indices.keys()):
            strongconnect(v)
    print(sccs)
    # 映射每个节点到所属的SCC
    node_to_scc = {}
    for i, scc in enumerate(sccs):
        for node in scc:
            node_to_scc[node] = i
    
    # 每个SCC的最长路径边数为节点数减一
    scc_lengths = [len(scc) - 1 for scc in sccs]
    
    # 构建缩点后的DAG和逆邻接表
    dag_adj = defaultdict(list)
    reverse_dag_adj = defaultdict(list)
    in_degree = defaultdict(int)
    
    # 添加所有SCC节点（确保孤立节点被包含）
    for scc_id in range(len(sccs)):
        dag_adj[scc_id]  # 确保键存在
    
    # 添加跨SCC的边
    for u, v in edge_list:
        scc_u = node_to_scc[u]
        scc_v = node_to_scc[v]
        if scc_u != scc_v and scc_v not in dag_adj[scc_u]:
            dag_adj[scc_u].append(scc_v)
            reverse_dag_adj[scc_v].append(scc_u)
            in_degree[scc_v] += 1
    
    # 确保所有SCC的入度初始化为0
    for scc_id in range(len(sccs)):
        if scc_id not in in_degree:
            in_degree[scc_id] = 0
    
    # Kahn算法拓扑排序
    queue = deque([scc for scc in dag_adj if in_degree[scc] == 0])
    topo_order = []
    while queue:
        u = queue.popleft()
        topo_order.append(u)
        for v in dag_adj[u]:
            in_degree[v] -= 1
            if in_degree[v] == 0:
                queue.append(v)
    
    # 动态规划计算最长路径
    dp = defaultdict(int)
    max_length = 0
    
    # 确保处理所有SCC（包括孤立节点）
    for scc in range(len(sccs)):
        if scc not in topo_order:
            topo_order.append(scc)
    
    for scc in topo_order:
        current_max = 0
        # 遍历所有前驱scc
        for predecessor in reverse_dag_adj.get(scc, []):
            current_max = max(current_max, dp[predecessor] + 1)
        
        dp[scc] = current_max + scc_lengths[scc]
        max_length = max(max_length, dp[scc])
    
    return max_length
```
